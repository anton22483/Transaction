```python
import os
import pandas as pd
from sqlalchemy import create_engine
```

```python
USER = 'postgres'
PASSWORD = 'root' 
HOST = 'localhost'
PORT = '5432'
DB_NAME = 'Transaction'
```

```python
connection_string = f'postgresql+psycopg://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB_NAME}'
engine = create_engine(connection_string)
print("Успех!!!")
```

```python
data_dir = 'data'

try:
    #Чтение
    df_clients = pd.read_csv(os.path.join(data_dir, 'clients.csv'))
    df_accounts = pd.read_csv(os.path.join(data_dir, 'accounts.csv'))
    df_transactions = pd.read_csv(os.path.join(data_dir, 'transactions.csv'))
    print("Прочитано!!!")
    
    df_clients['registration_date'] = pd.to_datetime(df_clients['registration_date']).dt.date
    df_accounts['open_date'] = pd.to_datetime(df_accounts['open_date']).dt.date
    df_transactions['transaction_date'] = pd.to_datetime(df_transactions['transaction_date'])
    
    #Отправка в DBeaver
    df_clients.to_sql('clients', con=engine, if_exists='append', index=False)
    df_accounts.to_sql('accounts', con=engine, if_exists='append', index=False)
    df_transactions.to_sql('transactions', con=engine, if_exists='append', index=False)
    
    print("Успех!!!")

except FileNotFoundError as e:
    print(f"Ошибка. Не найден файл.")
    print(f"Детали ошибки: {e}")
except Exception as e:
    print(f"Ошибка СУБД.")
    print(f"Детали ошибки: {e}")
```
